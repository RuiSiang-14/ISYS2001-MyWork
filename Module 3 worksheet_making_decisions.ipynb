### Activity 5: Build an admissions decision system

Universities weigh several factors to make admission decisions. Design and build a fair, logical version.

**Design first:**

1. Choose 3 to 4 factors (GPA, test scores, activities, essay, special circumstances)
2. Decide how they combine (all required? at least 2 of 3?)
3. Set specific thresholds
4. Think about edge cases and tie-breakers

> **Your criteria** (double-click to edit):

```
Factor 1: Test scores
Factor 2: weekly Activities marks
Factor 3: Attendance
Decision logic: I am calculating the Total marks of the students, the test scores is 50% of the total marks, while weekly activity marks is 30% of the total marks, attendance is 20% of the total marks.

Factors
Factor 1: Test Score – One test worth 100 marks.
Factor 2: Weekly Activities – 10 weekly activities, each worth 10 marks, for a total of 100 marks.
Factor 3: Attendance – 10 weeks of attendance, each week worth 10 marks, for a total of 100 marks.
Decision Logic

The total marks are calculated using the following weights:

Test Score = 50%
Weekly Activities = 30%
Attendance = 20%
```

````python
# Your admissions decision system
# Tip: for "at least 2 of 3 conditions true", try counting how many are True.
# This is my code

# Input marks
test_score = float(input("Enter the test score: "))
weekly_activity = float(input("Enter the weekly activity marks: "))
attendance = float(input("Enter the attendance marks: "))

# Calculate total marks
total_marks = (test_score * 0.50) + (weekly_activity * 0.30) + (attendance * 0.20)

# Display the result
print("Total Marks:", total_marks)

````

````python
# This is what chatgpt suggest

# Test Score (out of 100)
test_score = float(input("Enter Test Score (0-100): "))

# Weekly Activities (10 weeks, each out of 10 marks)
weekly_total = 0
for week in range(1, 11):
    while True:
        try:
            mark_input = input(f"Enter Week {week} Activity Mark (0-10): ")
            mark = float(mark_input)
            if 0 <= mark <= 10:
                break
            else:
                print("Please enter a mark between 0 and 10.")
        except ValueError:
            print("Invalid input. Please enter a number.")
    weekly_total += mark

# Convert weekly activities to percentage
weekly_score = weekly_total

# Attendance (10 weeks)
attendance_count = 0
for week in range(1, 11):
    while True:
        attendance_input = input(f"Was the student present in Week {week}? (Y/N): ")
        if attendance_input.upper() in ["Y", "N"]:
            break
        else:
            print("Invalid input. Please enter Y or N.")

    if attendance_input.upper() == "Y":
        attendance_count += 1

# Convert attendance to percentage
attendance_score = attendance_count * 10

# Calculate Total Marks
total_marks = (test_score * 0.50) + (weekly_score * 0.30) + (attendance_score * 0.20)

# Display Results
print("\n----- Student Result -----")
print("Test Score:", test_score)
print("Weekly Activity Score:", weekly_score)
print("Attendance Score:", attendance_score)
print("Total Marks:", total_marks)

# Final Decision
if total_marks >= 80:
    print("Result: Distinction")
elif total_marks >= 60:
    print("Result: Pass")
else:
    print("Result: Fail")
````

````python
# Test Score (out of 100)
test_score = float(input("Enter Test Score (0-100): "))

# Weekly Activities (10 weeks, each out of 10 marks)
weekly_total = 0
for week in range(1, 11):
    mark = float(input(f"Enter Week {week} Activity Mark (0-10): "))
    weekly_total += mark

# Convert weekly activities to percentage
weekly_score = weekly_total

# Attendance (10 weeks)
attendance_count = 0
for week in range(1, 11):
    attendance = input(f"Was the student present in Week {week}? (Y/N): ")

    if attendance.upper() == "Y":
        attendance_count += 1

# Convert attendance to percentage
attendance_score = attendance_count * 10

# Calculate Total Marks
total_marks = (test_score * 0.50) + (weekly_score * 0.30) + (attendance_score * 0.20)

# Display Results
print("\n----- Student Result -----")
print("Test Score:", test_score)
print("Weekly Activity Score:", weekly_score)
print("Attendance Score:", attendance_score)
print("Total Marks:", total_marks)

# Final Decision
if total_marks >= 80:
    print("Result: Distinction")
elif total_marks >= 60:
    print("Result: Pass")
else:
    print("Result: Fail")
````

## Key takeaways

**Computational thinking**
- Understand the problem before you code.
- Break decisions into clear rules.
- Test your logic with examples first.
- In an if/elif chain, order matters. Put the most specific conditions first.
- Use brackets to make complex conditions say what you mean. `and` binds tighter than `or`.

**Working with AI**
- AI is at its best when your requirements are clear.
- Use it to build your ideas, not to think for you.
- Give it your intent when you ask for help, especially when debugging.
