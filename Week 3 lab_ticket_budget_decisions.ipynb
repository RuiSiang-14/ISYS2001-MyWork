Budget decision

I am creating a budget decision to my customer, I will create a budget list for him on how to manage his monthly income

Factors
Factor 1: Food spending 15%
Factor 2: Hous loan or car loan 40%
Factor 3: Entertainment 10%
Factor 4: Daily expenses like oil, car park, toll 5%
Factor 5: Saving 30%

Decision logic
The percentage mean that he can't exceed the percentage of total income, if exceed print false, if there are money left after all calculation, classtification as extra saving. But, saving is different, the percentage means the minimum saving, if exceed 30% means good.


````python
# Paste the code your prompt produced here, then run it to show it works.

# Input monthly income
income = float(input("Enter monthly income: "))

# Input spending
food = float(input("Enter food spending: "))
loan = float(input("Enter house/car loan: "))
entertainment = float(input("Enter entertainment spending: "))
daily = float(input("Enter daily expenses: "))
saving = float(input("Enter savings amount: "))

# Calculate budget limits
food_limit = income * 0.15
loan_limit = income * 0.40
entertainment_limit = income * 0.10
daily_limit = income * 0.05
saving_limit = income * 0.30

# Check each category
if food > food_limit:
    print("Food Budget: Need to improve (Exceeded 15%)")
else:
    print("Food Budget: Good")

if loan > loan_limit:
    print("Loan Budget: Need to improve (Exceeded 40%)")
else:
    print("Loan Budget: Good")

if entertainment > entertainment_limit:
    print("Entertainment Budget: Need to improve (Exceeded 10%)")
else:
    print("Entertainment Budget: Good")

if daily > daily_limit:
    print("Daily Expenses Budget: Need to improve (Exceeded 5%)")
else:
    print("Daily Expenses Budget: Good")

if saving >= saving_limit:
    print("Saving Target: Good")
else:
    print("Saving Target: Need to improve (Below 30%)")

# Calculate money left
total_spending = food + loan + entertainment + daily + saving
extra_money = income - total_spending

print("\nTotal Spending:", total_spending)

if extra_money > 0:
    print("Extra Pocket Money:", extra_money)
elif extra_money == 0:
    print("No money left.")
else:
    print("Overspent by:", abs(extra_money))
````
