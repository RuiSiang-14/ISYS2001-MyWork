## Task 1: total sales by region, highest first

Load the file into a DataFrame, clean whatever needs cleaning so the sale values are
real numbers, and produce the total `Total_Sale` for each `Region`, ordered from
highest to lowest.

Plan it in plain words in the cell below before you write the pandas. If you use an
assistant to help with a line, give it your plan and make sure you can explain what
it hands back.

My plan, in plain words:

*(write your plan here before the code)*

```python
# Task 1: total sales by region, highest first.

#I will load the CSV file into pandas and check the data. Then I will remove the dollar sign from Total_Sale and convert it into numbers. After that, I will group the data by Region, add the Total_Sale for each region, and sort the results from highest to lowest
import pandas as pd

# Load the CSV file
df = pd.read_csv('sales_data.csv')

# Remove the dollar sign and convert Total_Sale to numbers
df['Total_Sale'] = df['Total_Sale'].str.replace('$', '', regex=False)
df['Total_Sale'] = pd.to_numeric(df['Total_Sale'])

# Calculate total sales for each region
sales_by_region = df.groupby('Region')['Total_Sale'].sum()

# Sort from highest to lowest
sales_by_region = sales_by_region.sort_values(ascending=False)

# Show the result
sales_by_region

```

## Task 2: units sold per product

Now find how many units of each product were sold in total, using the `Quantity`
column.

This is where looking at the data first pays off. If your numbers come out looking
like money rather than counts of units, stop and work out why before you go on. A
correct answer here should read as sensible unit counts, not dollar amounts.

My plan, in plain words:

```python
# Task 2: total units sold per product.

# I will clean the Quantity column by removing the dollar sign and converting it into numbers. Then I will group the data by Product and add the Quantity values to find the total number of units sold for each product
df['Quantity'] = df['Quantity'].str.replace('$', '', regex=False)
df['Quantity'] = pd.to_numeric(df['Quantity'])

df['Quantity']

units_by_product = df.groupby('Product')['Quantity'].sum()
units_by_product = units_by_product.sort_values(ascending=False)

units_by_product

```
